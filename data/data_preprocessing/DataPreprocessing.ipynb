```python
from sklearn.model_selection import train_test_split
from sklearn.preprocessing import LabelEncoder

l=LabelEncoder()
x=df.drop('pam50_+_claudin-low_subtype',axis=1)
y=df['pam50_+_claudin-low_subtype']
y_encoded=l.fit_transform(y)
x_train_base,x_test_base,y_train_base,y_test_base=train_test_split(
    x,
    y_encoded,
    test_size=0.20,
    random_state=42,
    stratify=y_encoded)
```

```python
# Check outliers in all 489 gene-expression features using IQR

outlier_summary = []

for gene in X_genes.columns:

    Q1 = X_genes[gene].quantile(0.25)
    Q3 = X_genes[gene].quantile(0.75)
    IQR = Q3 - Q1

    lower_bound = Q1 - 1.5 * IQR
    upper_bound = Q3 + 1.5 * IQR

    outliers = (
        (X_genes[gene] < lower_bound) |
        (X_genes[gene] > upper_bound)
    )

    outlier_count = outliers.sum()

    outlier_summary.append({
        'Gene': gene,
        'Q1': Q1,
        'Q3': Q3,
        'IQR': IQR,
        'Lower_Bound': lower_bound,
        'Upper_Bound': upper_bound,
        'Outlier_Count': outlier_count,
        'Outlier_Percentage': (outlier_count / len(X_genes)) * 100
    })

outlier_df = pd.DataFrame(outlier_summary)

outlier_df = outlier_df.sort_values(
    by='Outlier_Count',
    ascending=False
)

outlier_df.head(20)
```

```python
print("Total outlier observations across all genes:",
      outlier_df['Outlier_Count'].sum())

print("\nGenes having at least one outlier:",
      (outlier_df['Outlier_Count'] > 0).sum())

print("\nGenes having >5% outliers:",
      (outlier_df['Outlier_Percentage'] > 5).sum())

print("\nMaximum outlier percentage in any gene:",
      outlier_df['Outlier_Percentage'].max())
```

```python
from sklearn.feature_selection import f_classif

f_scores, p_values = f_classif(X_genes,y_train_base)

anova_results = pd.DataFrame({
    'Gene': X_genes.columns,
    'F_Score': f_scores,
    'P_Value': p_values
})

anova_results = anova_results.sort_values(
    by='F_Score',
    ascending=False
).reset_index(drop=True)

anova_results['Rank'] = anova_results.index + 1

anova_results.head(20)
```

```python
top_200_genes = anova_results.head(200)['Gene'].tolist()

X_genes = X_genes[top_200_genes]
```

```python
from xgboost import XGBClassifier
import numpy as np
from sklearn.preprocessing import LabelEncoder

xgb = XGBClassifier(n_estimators=100, random_state=42, eval_metric='mlogloss')

xgb.fit(X_genes, y_train_base)

importances = xgb.feature_importances_
top_50_indices = np.argsort(importances)[::-1][:50]

# 3. Extract the final top 50 genomic/mutation feature names
final_top_50_genomic_cols = X_genes.columns[top_50_indices]

# print("Final Top 50 Genomic Features Count:", len(final_top_50_genomic_cols))
print("Top 50 Features:", list(final_top_50_genomic_cols))
```

```python
mut_cols=[]
for c in final_top_50_genomic_cols:
    if str(c)+"_mut" in x_train_base.columns:
        mut_cols.append(str(c)+"_mut")
```

```python
for col in mut_cols:
    x_train_base[col] = x_train_base[col].apply(
        lambda x: 0 if str(x).strip() == "0" else 1
    )
    x_test_base[col] = x_test_base[col].apply(
        lambda x: 0 if str(x).strip() == "0" else 1
    )
```

```python
x_train_clinical=x_train_base.iloc[:,:21]
x_train_genes=x_train_base[final_top_50_genomic_cols]
x_train_mut=x_train_base[mut_cols]
y_train=l.inverse_transform(y_train_base)
train_dataset_temp=pd.concat([x_train_clinical,x_train_genes,x_train_mut],axis=1)
train_dataset_temp['pam50_+_claudin-low_subtype']=y_train
train_dataset_temp.head()
```

```python
x_test_clinical=x_test_base.iloc[:,:21]
x_test_genes=x_test_base[final_top_50_genomic_cols]
x_test_mut=x_test_base[mut_cols]
y_test=l.inverse_transform(y_test_base)
test_dataset=pd.concat([x_test_clinical,x_test_genes,x_test_mut],axis=1)
test_dataset['pam50_+_claudin-low_subtype']=y_test
test_dataset.head()
```

```python


train_dataset, validation_dataset = train_test_split(
    train_dataset_temp,
    test_size=0.125,
    random_state=42,
    stratify=train_dataset_temp['pam50_+_claudin-low_subtype']
)

print("Final training shape:", train_dataset.shape)
print("Validation shape:", validation_dataset.shape)
print("Test shape:", test_dataset.shape)
```
